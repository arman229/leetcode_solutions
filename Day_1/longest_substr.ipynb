## Longest Substring Without Repeating Characters   

---

Given a string `s`, find the length of the longest substring without duplicate characters.  

---

### Example 1
**Input:**  
`s = "abcabcbb"`  

**Output:**  
`3`  

**Explanation:**  
The answer is `"abc"`, with the length of `3`.  

---

### Example 2
**Input:**  
`s = "bbbbb"`  

**Output:**  
`1`  

**Explanation:**  
The answer is `"b"`, with the length of `1`.  

---

### Example 3
**Input:**  
`s = "pwwkew"`  

**Output:**  
`3`  

**Explanation:**  
The answer is `"wke"`, with the length of `3`.  
Notice that the answer must be a **substring**.  
`"pwke"` is a **subsequence**, not a substring.  

---

### Constraints
- `0 <= s.length <= 5 * 10^4`  
- `s` consists of English letters, digits, symbols, and spaces.  


```python
class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        mapping_set = set()
        max_len:int = 0
        left_side:int = 0
        for i in range(len(s)):
            while s[i] in mapping_set:
                mapping_set.remove(s[left_side])
                left_side +=1
            mapping_set.add(s[i])
            if i - left_side + 1 > max_len:
                max_len = i-left_side + 1
        return max_len
    
  

sol = Solution()                    
print(sol.lengthOfLongestSubstring('abceda'))
```

### Complexity Analysis 
- Time Complexity: O(n)  
- Space Complexity: O(m)  

```python
class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        left = right = 0
        find_len = 0
        prev_index = {}
        while right < len(s):
            if s[right] in prev_index and prev_index[s[right]] >= left:
                left = prev_index[s[right]] + 1
            prev_index[s[right]] = right
            find_len = max(find_len, right-left+1)
            right +=1
        return find_len
sol = Solution()
print(sol.lengthOfLongestSubstring('aabbaa'))

    
```
