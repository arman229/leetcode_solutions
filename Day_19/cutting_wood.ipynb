## Cutting Wood

---

You are given an array `heights` representing the heights of trees, and an integer `k` representing the total length of wood that needs to be collected.

A wood-cutting machine is set to a certain height `H`.

- Any tree taller than `H` is cut down to height `H`.
- The wood collected from a tree is equal to the portion above `H`:
  ```text
  tree_height - H
  ```
- Trees with height less than or equal to `H` are left untouched.

Determine the **highest possible value of `H`** such that the total amount of wood collected is **at least `k`**.

Assume the wood-cutting machine cannot be set higher than the tallest tree in the array.

---

### Example

**Input:**
```text
heights = [2, 6, 3, 8]
k = 7
```

**Output:**
```text
3
```

**Explanation:**

If the cutting height is `H = 3`:

| Tree Height | Wood Collected |
|-------------|---------------:|
| 2 | 0 |
| 6 | 3 |
| 3 | 0 |
| 8 | 5 |

Total wood collected:

```text
0 + 3 + 0 + 5 = 8
```

Since `8 >= 7`, the cutting height `H = 3` satisfies the requirement.

Any height greater than `3` would collect less wood, making `3` the **highest valid cutting height**.

---

### Constraints

- It is always possible to collect at least `k` meters of wood.
- There is at least one tree.
- `1 <= heights.length`
- `1 <= heights[i]`
- `1 <= k`
```

````python
def find_height(mid,heights,k):
    total_len = 0
    for he in heights:
        if he > mid:
            total_len += he-mid
    return total_len>=k
def cutting_wood(heights, k): 
    left, right = 1, max(heights)
    while left < right:
        mid = (left + right)//2 + 1
        if find_height(mid,heights,k):
            left = mid
        else:
            right = mid -1
    return left
cutting_wood([2, 6, 3, 8], 7)
````

### Complexity Analysis 
- Time Complexity: O(nlog(m))    where n denotes the number of trees, and m denotes the maximum height of the trees. 
- Space Complexity: O(1)
