 
## 2454. Next Greater Element IV

 

---

You are given a **0-indexed** array of non-negative integers `nums`. For each integer in `nums`, you must find its respective **second greater** integer.

The **second greater** integer of `nums[i]` is `nums[j]` such that:

- `j > i`
- `nums[j] > nums[i]`
- There exists **exactly one** index `k` such that `nums[k] > nums[i]` and `i < k < j`.

If there is no such `nums[j]`, the second greater integer is considered to be `-1`.

### Example

In the array:

```text
[1, 2, 4, 3]
````

* The second greater integer of `1` is `4`.
* The second greater integer of `2` is `3`.
* The second greater integer of `3` is `-1`.
* The second greater integer of `4` is `-1`.

Return an integer array `answer`, where `answer[i]` is the second greater integer of `nums[i]`.

---

### Example 1

**Input:**

```text
nums = [2,4,0,9,6]
```

**Output:**

```text
[9,6,6,-1,-1]
```

**Explanation:**

* **Index 0:** `4` is the first integer greater than `2`, and `9` is the second integer greater than `2`.
* **Index 1:** `9` is the first integer greater than `4`, and `6` is the second integer greater than `4`.
* **Index 2:** `9` is the first integer greater than `0`, and `6` is the second integer greater than `0`.
* **Index 3:** There is no integer greater than `9` to its right, so the second greater integer is `-1`.
* **Index 4:** There is no integer greater than `6` to its right, so the second greater integer is `-1`.

Therefore:

```text
answer = [9,6,6,-1,-1]
```

---

### Example 2

**Input:**

```text
nums = [3,3]
```

**Output:**

```text
[-1,-1]
```

**Explanation:**

Neither integer has any integer greater than it to its right, so the result is:

```text
[-1,-1]
```

---

### Constraints

* `1 <= nums.length <= 10^5`
* `0 <= nums[i] <= 10^9`

```


`````python
from typing import List
class Solution:
    def secondGreaterElement(self, nums: List[int]) -> List[int]:
        n = len(nums)
        resp:List[int] = [-1]*n
        stack1:List[int] = []
        stack2:List[int] = []
        stack3:List[int] = []
        
        for i in range(n):
            while stack1 and nums[stack1[-1]] < nums[i]:
                stack3.append(stack1.pop())
            while stack2 and nums[stack2[-1]] < nums[i]:
                index = stack2.pop()
                resp[index] = nums[i]
            while stack3:
                stack2.append(stack3.pop())
            stack1.append(i)
        return resp
`````

### Complexity Analysis 
- Time Complexity: O(n)  
- Space Complexity: O(n)
