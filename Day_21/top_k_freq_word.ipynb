 
## 692. Top K Frequent Words


---

Given an array of strings `words` and an integer `k`, return the **`k` most frequent strings**.

Return the answer **sorted** by:

1. **Frequency** from highest to lowest.
2. Words with the same frequency are sorted by their **lexicographical order**.

---

### Example 1

**Input:**

```text
words = ["i","love","leetcode","i","love","coding"], k = 2
````

**Output:**

```text
["i","love"]
```

**Explanation:**

`"i"` and `"love"` are the two most frequent words.

Note that `"i"` comes before `"love"` due to its lower alphabetical order.

---

### Example 2

**Input:**

```text
words = ["the","day","is","sunny","the","the","the","sunny","is","is"], k = 4
```

**Output:**

```text
["the","is","sunny","day"]
```

**Explanation:**

The four most frequent words are `"the"`, `"is"`, `"sunny"`, and `"day"`, with the number of occurrences being:

```text
the   → 4
is    → 3
sunny → 2
day   → 1
```

---

### Constraints

* `1 <= words.length <= 500`
* `1 <= words[i].length <= 10`
* `words[i]` consists of lowercase English letters.
* `k` is in the range `[1, number of unique words]`

---

### Follow-up

Could you solve it in **O(n log(k))** time and **O(n)** extra space?

```

`````python
from typing import List
class Solution:
    class Pair:
        def __init__(self, string, freq):
            self.string = string
            self.freq = freq
    @staticmethod
    def higher_priority(a, b):
        if a.freq != b.freq:
            return a.freq > b.freq
        return a.string < b.string

    @staticmethod
    def heapify_down(heap, index):
        n = len(heap)
        while True:        
            largest = index
            left = 2 * index + 1
            right = 2 * index + 2
            if left < n and Solution.higher_priority(heap[left], heap[largest]):
                largest = left
            if right < n and Solution.higher_priority(heap[right], heap[largest]):
                largest = right
            if largest == index:
                break
            heap[index], heap[largest] = heap[largest], heap[index]
            index = largest

    @staticmethod
    def heapify(heap):
        n = len(heap)
        start = (n // 2) - 1
        for i in range(start, -1, -1):
            Solution.heapify_down(heap, i)

    @staticmethod
    def heap_pop(heap):
        if len(heap) == 0:
            return None

        if len(heap) == 1:
            return heap.pop()

        top = heap[0]

        heap[0] = heap.pop()

        Solution.heapify_down(heap, 0)

        return top

    @staticmethod
    def count_frequency(strings):
        freq = {}
        for string in strings:
            freq[string] = freq.get(string, 0) + 1 
        return freq
    
    def topKFrequent(self, words: List[str], k: int) -> List[str]:
        freqs = Solution.count_frequency(words)
        heap = []
        for string, freq in freqs.items():
            heap.append(Solution.Pair(string, freq))
        Solution.heapify(heap)
        
        result = []
        for _ in range(k):
            pair = Solution.heap_pop(heap)
            result.append(pair.string)
        return result
`````

### Complexity Analysis 
Let n = total number of words, m = number of unique words, and k = number of words to return.
- Time Complexity: O(n + k log m)
- Space Complexity: O(m)  

`````python
from collections import Counter
import heapq
class Solution:
    class Pair:
        def __init__(self, string, freq):
            self.string = string
            self.freq = freq 
        def __lt__(self, other):  
            if self.freq== other.freq: 
                return self.string < other.string 
            return self.freq> other.freq 
    def topKFrequent(self, words: List[str], k: int) -> List[str]:
        freqs = Counter(words) 
        max_heap = [Solution.Pair(str, freq) for str, freq in freqs.items()] 
        heapq.heapify(max_heap) 
        return [heapq.heappop(max_heap).string for _ in range(k)] 
`````
