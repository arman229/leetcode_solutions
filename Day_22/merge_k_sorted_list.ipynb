## 23. Merge k Sorted Lists


---

You are given an array of `k` linked lists `lists`, where each linked list is sorted in **ascending order**.

Merge all the linked lists into **one sorted linked list** and return it.

---

### Example 1

**Input:**

```text
lists = [[1,4,5],[1,3,4],[2,6]]
````

**Output:**

```text
[1,1,2,3,4,4,5,6]
```

**Explanation:**

The linked lists are:

```text
1 -> 4 -> 5
1 -> 3 -> 4
2 -> 6
```

Merging them produces one sorted linked list:

```text
1 -> 1 -> 2 -> 3 -> 4 -> 4 -> 5 -> 6
```

---

### Example 2

**Input:**

```text
lists = []
```

**Output:**

```text
[]
```

---

### Example 3

**Input:**

```text
lists = [[]]
```

**Output:**

```text
[]
```

---

### Constraints

* `k == lists.length`
* `0 <= k <= 10^4`
* `0 <= lists[i].length <= 500`
* `-10^4 <= lists[i][j] <= 10^4`
* `lists[i]` is sorted in **ascending order**.
* The sum of `lists[i].length` will not exceed `10^4`.

```


`````python
# Definition for singly-linked list.
from typing import List, Optional
import heapq
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
    def __lt__(self, other):
        return self.val < other.val 
class Solution:
    def mergeKLists(self, lists: List[Optional[ListNode]]) -> Optional[ListNode]:
        heap = []
        for i,head in enumerate(lists): 
            if head:
                heapq.heappush(heap, ( head.val,i,head)) 
                
        dummy = ListNode(-1)
        curr = dummy
        while heap:
            val,i, smallest_node = heapq.heappop(heap)
            curr.next = smallest_node
            if smallest_node.next:
                heapq.heappush(heap,(smallest_node.next.val,i,smallest_node.next))
            curr = curr.next
        return dummy.next
        
`````

### Complexity Analysis 
- Time Complexity: O(nlog(k))  where k = number of linked lists, N = total number of nodes across all lists
- Space Complexity: O(k)
