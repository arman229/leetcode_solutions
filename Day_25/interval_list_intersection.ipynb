## 986. Interval List Intersections


---

You are given two lists of closed intervals, `firstList` and `secondList`, where:

```text
firstList[i] = [starti, endi]
secondList[j] = [startj, endj]
```

Each list of intervals is **pairwise disjoint** and in **sorted order**.

Return the **intersection** of these two interval lists.

A **closed interval** `[a, b]` (with `a <= b`) denotes the set of real numbers `x` such that:

```text
a <= x <= b
```

The **intersection** of two closed intervals is a set of real numbers that is either empty or represented as a closed interval.

For example, the intersection of `[1,3]` and `[2,4]` is `[2,3]`.

---

### Example 1

**Input:**

```text
firstList = [[0,2],[5,10],[13,23],[24,25]]
secondList = [[1,5],[8,12],[15,24],[25,26]]
```

**Output:**

```text
[[1,2],[5,5],[8,10],[15,23],[24,24],[25,25]]
```

---

### Example 2

**Input:**

```text
firstList = [[1,3],[5,9]]
secondList = []
```

**Output:**

```text
[]
```

---

### Constraints

* `0 <= firstList.length, secondList.length <= 1000`
* `firstList.length + secondList.length >= 1`
* `0 <= starti < endi <= 10^9`
* `endi < starti + 1`
* `0 <= startj < endj <= 10^9`
* `endj < startj + 1`


````python
from typing import List
class Solution:
    def intervalIntersection(self, firstList: List[List[int]], secondList: List[List[int]]) -> List[List[int]]:
        i=j=0
        merge = []
        
        while i < len(firstList) and j < len(secondList):
            if firstList[i][0] <= secondList[j][0]:
                A = firstList[i]
                B = secondList[j]
            else:
                A = secondList[j]
                B = firstList[i]
            if B[0]<=A[-1]:
                merge.append([B[0],min(A[-1],B[-1])])
            if firstList[i][-1] < secondList[j][-1]:
                i +=1
            else:
                j +=1
        return merge 

````

### Complexity Analysis 
- Time Complexity: O(n)
- Space Complexity: O(n)
