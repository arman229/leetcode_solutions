# Largest Overlap of Intervals

Given an array of intervals, determine the **maximum number of intervals that overlap at any point**.

Each interval is **half-open**, meaning it includes the **start point** but excludes the **end point**.

### Example

**Input:**

```text
intervals = [[1,3), [2,6), [4,8), [6,7), [5,7)]
```

**Output:**

```text
3
```

### Explanation

The maximum number of overlapping intervals at any point is **3**.

### Constraints

* The input will contain at least one interval.
* For every index `i` in the array:

```text
intervals[i].start < intervals[i].end
```


````python
Interval = [[1,2], [2,6], [4,8],[6,7],[5,7]]
def largest_overlap_of_intervals(intervals) -> int: 
    points = []
    active_intervals = 0
    max_count = 0
    for i in Interval:
        points.append((i[0],'S'))
        points.append((i[1],'E'))
    
    points.sort(key=lambda x: (x[0], x[1]))
    for val, type in points:
        if type == 'S':
            active_intervals += 1
        else:
            active_intervals -= 1
        max_count = max(max_count, active_intervals)
    return max_count
largest_overlap_of_intervals(Interval)  

        
        
````

### Complexity Analysis 
- Time Complexity: O(nlog(n))
- Space Complexity: O(n)
