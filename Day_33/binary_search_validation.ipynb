 
# 98. Validate Binary Search Tree

Given the `root` of a binary tree, **determine if it is a valid Binary Search Tree (BST)**.

A **valid BST** is defined as follows:

- The **left subtree** of a node contains only nodes with keys **strictly less than** the node's key.
- The **right subtree** of a node contains only nodes with keys **strictly greater than** the node's key.
- Both the left and right subtrees must also be **binary search trees**.

**Output:**

## Example 2

![Validate Binary Search Tree](https://assets.leetcode.com/uploads/2020/12/01/tree2.jpg)

**Input:**

```text
root = [5,1,4,null,null,3,6]
```

**Output:**

```text
false
```

**Explanation:**

The root node's value is `5`, but its right child's value is `4`.

Since `4 < 5`, this violates the BST rule that every node in the right subtree must have a value **strictly greater than** the root.

---

## Constraints

* The number of nodes in the tree is in the range `[1, 10^4]`.
* `-2^31 <= Node.val <= 2^31 - 1`

```


`````python
from collections import deque
from typing import Optional

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
        
class Solution:
    def isValidBST(self, root: Optional[TreeNode]) -> bool:
        
        if not root:
            return False
        queue =  deque([(root, float("-inf"), float("inf"))])
        while queue:
            node,leftrange,rightrange = queue.popleft()
            if not (leftrange < node.val < rightrange):
                return False
            if node.left:
                queue.append((node.left,leftrange,node.val))
            if node.right:
                queue.append((node.right,node.val,rightrange))
        return True
`````

### Complexity Analysis 
- Time Complexity: O(n)
- Space Complexity: O(n)
