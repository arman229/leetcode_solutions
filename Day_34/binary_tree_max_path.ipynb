# 124. Binary Tree Maximum Path Sum

A **path** in a binary tree is a sequence of nodes where each pair of adjacent nodes in the sequence has an edge connecting them. A node can only appear in the sequence **at most once**.

> The path does not need to pass through the root.

The **path sum** of a path is the sum of the node's values in the path.

Given the `root` of a binary tree, return *the maximum **path sum** of any **non-empty** path*.

---

## Example 1

![Binary Tree Maximum Path Sum](https://assets.leetcode.com/uploads/2020/10/13/exx1.jpg)

**Input:**

```text
root = [1,2,3]
````

**Output:**

```text
6
```

**Explanation:**

The optimal path is:

```text
2 -> 1 -> 3
```

The path sum is:

```text
2 + 1 + 3 = 6
```

## Example 2

![Binary Tree Maximum Path Sum](https://assets.leetcode.com/uploads/2020/10/13/exx2.jpg)

**Input:**

```text
root = [-10,9,20,null,null,15,7]
```

**Output:**

```text
42
```

**Explanation:**

The optimal path is:

```text
15 -> 20 -> 7
```

The path sum is:

```text
15 + 20 + 7 = 42
```

---

## Constraints

* The number of nodes in the tree is in the range `[1, 3 * 10^4]`.
* `-1000 <= Node.val <= 1000`

```


`````python
from typing import Optional
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
class Solution:
    def __init__(self):
        self.max_sum = float("-inf")
    
    def findpathfun(self, root):
        if root is None:
            return 0
        find_left = max(self.findpathfun(root.left),0)
        find_right = max(self.findpathfun(root.right),0)
        self.max_sum = max(self.max_sum,root.val+find_left+find_right)
        return root.val + max(find_left,find_right)

    def maxPathSum(self, root: Optional[TreeNode]) -> int:
        self.findpathfun(root)
        return self.max_sum
`````

### Complexity Analysis 
- Time Complexity: O(n)
- Space Complexity: O(n)
